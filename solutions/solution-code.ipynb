## Review of Linear Regression
---

Linear regression is a way for us to relate some dependent variable $y$ to independent variables $x_1, \dots, x_p$. We can write it out like this:

$$y_i = \beta_0 + \beta_1x_1 + \cdots + \beta_px_p + e_i$$

or if you're fancy, via matrices like this:

$$y_i = \mathbf{x}_i^T\beta + e_i$$

#### LINE Assumptions
We need to make four assumptions in order for linear regression to work "properly". They are:

1. **L**inearity: The relationship between $y$ and the $x$s is linear.
1. **I**ndependence: Each observation is independent from one another.
1. **N**ormality: The errors (residuals) follow a normal distribution.
1. **E**quality of variance: The errors (residuals) seem to follow the same variance regardless of $\hat{y}$.

We can measure the performance of our model by using mean squared error (MSE).

## Feature Engineering
---

> "Coming up with features is difficult, time-consuming, requires expert knowledge. 'Applied machine learning' is basically feature engineering." - Andrew Ng 

**Feature engineering** is a broad term to mean the creation and manipulation of features (variables) for use in machine learning algorithms.

* **Sometimes it's helpful.** For example, if we transform an $x$-variable using some function (maybe a log, maybe a square), we get a better model.
* **Sometimes it's mandatory.** We'll see ML algorithms where scaling $x$-variables is necessary for the algorithm to work properly. As another example, we'll need to preprocess and engineer text data for use in NLP models.
* **Sometimes it's automatic.** There are algorithms for automatically engineering features according to some naive rules. Notably, neural nets do this to a minor extent.

Suffice it to say: If your features (variables) aren't good, your predictions and inferences won't be good!

In general, unless we're working with the same data over and over again, this isn't something we can automate. It will require creativity and a good, thorough understanding of our data.

````python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OneHotEncoder
from sklearn.compose import ColumnTransformer
````

## Load the Data

---

Drop `Unnamed: 0` once you've loaded the csv into a `DataFrame`.

````python
ads = pd.read_csv('datasets/Advertising.csv')
ads.drop(columns='Unnamed: 0', inplace=True)
ads.head()
````

## Create our features matrix (`X`) and target vector (`y`)
---

The following columns will be our features:
- `TV`
- `radio`
- `newspaper`

The `sales` column is our target; the column we're trying to predict.

In the cell below, create your `X` and `y` variables.

````python
xvars = ['TV', 'radio', 'newspaper']
X = ads[xvars]
y = ads['sales']
````

---

````python
lr = LinearRegression()
````

---

# Interaction terms
---

Imagine seeing a commercial on TV, and then hearing a commercial by the same company a few days later. If you eventually purchase the product in question, was it the TV spot that swayed you or the radio commercial...or was a it a synergy of the two commercials that did the trick?

**Interaction terms** allow us to see what affect the **combination** of two features have on our label. For example, I'm not a fan of plain hot dogs, and I never eat mustard by itself. But I do enjoy hot dogs **with** mustard.

To put another way:

> If we believe $y$ varies with $x_1$ differently depending on the value of another variable $x_2$, we might wish to include the interaction term of $x_1x_2$.

In the cell below, create an interaction column between `TV` and `radio`.

````python
ads['tv_x_radio'] = ads['TV'] * ads['radio']
````

## EDA: Create a heatmap
---

In the cell below, create a heatmap in `seaborn` to show how our `TV * radio` interaction term correlates to all the other columns.

````python
sns.heatmap(ads.corr(), annot=True, cmap='coolwarm')
````

## EDA: Pairplot
---

Now use `seaborn`'s `.pairplot()` method to create scatterplots with our new column.

````python
sns.pairplot(ads);
````

## Create our features matrix (`X`) and target vector (`y`)
---

Our new column is highly correlated to our label, so let's incorporate it in our features matrix:
- `TV`
- `radio`
- `newspaper`
- `TV * radio` interaction term

The `sales` column is our label: the column we're trying to predict.

In the cell below, create your `X` and `y` variables.

````python
X = ads.drop('sales', axis=1)
X.head()
````

## Model evaluation
---

Use `cross_val_score` to see how much our new feature moves the $R^2$ score.

````python
cross_val_score(lr, X, y, cv=5).mean()
````

## Manually Transforming a new Feature
But wait! There's more! If we look at the pairplot above, we see that the L assumption in LINE might be violated for TV sales. Why not try including $x_{TV}^2$ in our model instead of an interaction term?

Let's try that below:

````python
ads['tv2'] = ads['TV']**2
X = ads[['TV', 'tv2', 'radio', 'newspaper', 'tv_x_radio']]

lr = LinearRegression()
cross_val_score(lr, X, y, cv=5).mean()
````

````python
# Let's recreate our X and y for the ad data's original features
X = ads[['TV', 'radio', 'newspaper']]
y = ads['sales']
````

# `PolynomialFeatures`
We'll use the `PolynomialFeatures` transformer on all of our numeric variables to create this data.

````python
poly = PolynomialFeatures(include_bias=False)
````

### Aside: Estimators vs Transformers
Scikit-learn has two types of objects for us to work with: estimators and transformers. Here's the difference:

#### Estimators
Estimators are models. They take input (`X`, `y`) and make predictions (_estimates_). For example, `LinearRegression()` is an estimator.

> Estimators have `.fit()` and `.predict()` methods.

#### Transformers
Transformers are what they sound like: they transform the data. They take in data matrices (`X`) and transform them into new data matrices.

> Estimators have `.fit()` and `.transform()` methods. These are used in conjunction so often that they also have a `.fit_transform()` method.

````python
poly.fit(X)
X_poly = poly.transform(X)
````

````python
poly.get_feature_names(X.columns)
````

````python
X_poly.shape
````

````python
# Let's fit our model with both X and X_poly and see the difference.
lm = LinearRegression()
print(cross_val_score(lm, X, y, cv=5).mean())
print(cross_val_score(lm, X_poly, y, cv=5).mean())
````

# `OneHotEncoder`: New name, familiar game
Another name for dummifying variables is "one-hot encoding" them. The term comes from the computer science side of things, but it definitely sounds cooler.

````python
cat_vars = ['month', 'day']
X = fire[cat_vars]
oh = OneHotEncoder(drop='first')
X_oh = oh.fit_transform(X)
````

````python
X_oh.toarray()
````

# `ColumnTransformer`: Bringing it all together
The `ColumnTransformer` transformer aggregates multiple other transformers to create one big transformer. You may have noticed that the ones we've seen so far (`StandardScaler` and `PolynomialFeatures`) only operate on an entire matrix at once. `ColumnTransformer` is how we get around that!

The individual transformers are given as arguments in this order:

```
(name, transformer, columns)
```

````python
ct = ColumnTransformer([
    ('sc', StandardScaler(), ['X', 'Y']),
    ('pf', PolynomialFeatures(), ['FFMC', 'DMC', 'DC', 'ISI', 'temp', 'RH', 'wind', 'rain']),
    ('oh', OneHotEncoder(drop='first', sparse=False), ['month', 'day'])
])

X_ct = ct.fit_transform(fire)
````
